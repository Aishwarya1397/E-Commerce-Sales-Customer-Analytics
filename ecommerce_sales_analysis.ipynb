```python
previous_purchase_analysis = (
    df.groupby("Previous Purchases")["Purchase Amount (USD)"]
      .agg(["count", "sum", "mean", "median", "std"])
      .reset_index()
)

previous_purchase_analysis
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(previous_purchase_analysis["Previous Purchases"],
         previous_purchase_analysis["mean"],
         marker="o")

plt.title("Average Purchase Amount vs Previous Purchases")
plt.xlabel("Previous Purchases")
plt.ylabel("Average Purchase Amount (USD)")
plt.grid(True)

plt.show()
```

```python
bins = [15,25,35,45,55,65,75]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "66-75"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```
